```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
data = np.array([
    [1, 2],
    [2, 3],
    [5, 5],
    [6, 4]
])

names = ["P1", "P2", "P3", "P4"]

df = pd.DataFrame(data, columns=["X", "Y"], index=names)

print("Dataset:")
display(df)
```

```python
n = len(data)

D = np.zeros((n, n))

for i in range(n):
    for j in range(i + 1, n):
        distance = np.sqrt(
            (data[i, 0] - data[j, 0])**2 +
            (data[i, 1] - data[j, 1])**2
        )

        D[i, j] = distance
        D[j, i] = distance

print("Dissimilarity Matrix (Manual):")
print(np.round(D, 3))
```

```python
from sklearn.metrics import pairwise_distances

D_library = pairwise_distances(
    data,
    metric="euclidean"
)

print("Dissimilarity Matrix (Library):")
print(np.round(D_library, 3))
```

```python
distance_df = pd.DataFrame(
    D_library,
    index=names,
    columns=names
)

print("Euclidean Dissimilarity Matrix:")
display(distance_df.round(3))
```

```python
plt.figure(figsize=(6, 5))

sns.heatmap(
    distance_df,
    annot=True,
    fmt=".3f",
    cmap="YlGnBu"
)

plt.title("Dissimilarity Matrix - Euclidean Distance")
plt.tight_layout()
plt.show()
```
